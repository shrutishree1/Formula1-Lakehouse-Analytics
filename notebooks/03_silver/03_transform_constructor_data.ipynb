#### Steps taken:
1. Read bronze constructor table.
2. Keep only columns required for analytics, will drop the url column.
3. Standardize the column name using snake case (ex: constructorID as constructor_is)
4. Renaming columns to make it more meaningful (ex: name to construtor_name).
5. Remove duplicate records.
6. Transform values of columns to title case (ex: nationality).
7. Write the transform data to silver constructor table.

```python
%run ../environment_config
```

```python
bronze_table = f"{catalog}.{bronze_schema}.constructors"
silver_table = f"{catalog}.{silver_schema}.constructors"
```

```python
constructor_df = spark.read.table(bronze_table)
```

```python
constructor_selected_df = constructor_df.drop(F.col("url"))
```

```python
display(constructor_selected_df)
```

```python
constructor_rename_df = constructor_selected_df.withColumnsRenamed({
                                        "constructorId" : "constructor_id",
                                        "name" : "constructor_name" 
                                       })
```

```python
display(constructor_rename_df)
```

```python
constructor_distinct_df = constructor_rename_df.dropDuplicates(["constructor_id"])
```

```python
constructor_final_df = constructor_distinct_df.withColumn( 
                                            "nationality" , F.initcap(F.col("nationality"))
)
```

```python
(
    constructor_final_df.write
                        .format("delta")
                        .mode("overwrite")
                        .saveAsTable(silver_table)
)
```
